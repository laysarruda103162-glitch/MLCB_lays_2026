```python
# Importando as bibliotecas necessárias

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Base de dados para treinamento

frases = [
    "Eu adorei o atendimento",
    "O produto é excelente",
    "Estou muito satisfeito com a compra",
    "A entrega foi rápida e perfeita",
    "Gostei muito do serviço",
    "O atendimento foi ótimo",
    "Recomendo este produto",
    "Minha experiência foi maravilhosa",
    "O produto chegou em perfeito estado",
    "Estou feliz com a compra",

    "Eu odiei o atendimento",
    "O produto é péssimo",
    "Estou muito insatisfeito com a compra",
    "A entrega demorou demais",
    "Não gostei do serviço",
    "O atendimento foi horrível",
    "Não recomendo este produto",
    "Minha experiência foi muito ruim",
    "O produto chegou quebrado",
    "Estou decepcionado com a compra"
]

sentimentos = [
    "positivo", "positivo", "positivo", "positivo", "positivo",
    "positivo", "positivo", "positivo", "positivo", "positivo",
    "negativo", "negativo", "negativo", "negativo", "negativo",
    "negativo", "negativo", "negativo", "negativo", "negativo"
]
```

```python
# Separando os dados em treino e teste

X_treino, X_teste, y_treino, y_teste = train_test_split(
    frases,
    sentimentos,
    test_size=0.30,
    random_state=42,
    stratify=sentimentos
)

print("Frases para treinamento:", len(X_treino))
print("Frases para teste:", len(X_teste))
```

```python
# Transformando os textos em números com CountVectorizer

vetorizador = CountVectorizer()

X_treino_vetorizado = vetorizador.fit_transform(X_treino)
X_teste_vetorizado = vetorizador.transform(X_teste)

print("Quantidade de palavras aprendidas:", len(vetorizador.get_feature_names_out()))
print("Algumas palavras:", vetorizador.get_feature_names_out()[:10])
```

```python
# Treinando o modelo de Machine Learning

modelo = MultinomialNB()

modelo.fit(X_treino_vetorizado, y_treino)

print("Modelo treinado com sucesso!")
```

```python
# Fazendo previsões e avaliando o modelo

previsoes = modelo.predict(X_teste_vetorizado)

acuracia = accuracy_score(y_teste, previsoes)

print("Acurácia do modelo:", acuracia)
print("\nResultados:")

for frase, real, previsto in zip(X_teste, y_teste, previsoes):
    print(f"Frase: {frase}")
    print(f"Real: {real} | Previsto: {previsto}")
    print("-" * 40)
```

```python
# Testando o modelo com uma frase nova

frase_nova = ["O atendimento foi excelente"]

frase_vetorizada = vetorizador.transform(frase_nova)

resultado = modelo.predict(frase_vetorizada)

print("Frase:", frase_nova[0])
print("Sentimento previsto:", resultado[0])
```

```python
# Testando uma frase negativa

frase_nova = ["O atendimento foi horrível"]

frase_vetorizada = vetorizador.transform(frase_nova)

resultado = modelo.predict(frase_vetorizada)

print("Frase:", frase_nova[0])
print("Sentimento previsto:", resultado[0])
```
